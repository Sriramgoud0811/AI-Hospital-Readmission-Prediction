### ✂️ 1.1.2 Train-Test Split:

### Case Scenario

The model needs one portion of historical encounters to learn patterns and a separate portion to evaluate how well it performs on unseen patients. Because some patients have multiple encounters, we will keep all encounters belonging to the same patient in only one split.

### 👨‍💻 Code

We need patient_nbr only for creating the groups. It is not an X feature.

```python
from sklearn.model_selection import GroupShuffleSplit

# Patient IDs are used only to keep the same patient in one split
groups = df_model["patient_nbr"]

# Create an 80:20 patient-aware train-test split
group_split = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=42
)

train_idx, test_idx = next(
    group_split.split(X, y, groups=groups)
)

# Create train and test datasets
X_train = X.iloc[train_idx].copy()
X_test = X.iloc[test_idx].copy()

y_train = y.iloc[train_idx].copy()
y_test = y.iloc[test_idx].copy()
```

```python
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

```python
X_train.shape[0] / X.shape[0], X_test.shape[0] / X.shape[0]
```

```python
train_patients = set(df_model.iloc[train_idx]["patient_nbr"])
test_patients = set(df_model.iloc[test_idx]["patient_nbr"])

len(train_patients.intersection(test_patients))
```

## 0 means no patient appears in both training and test data.

## I use patient_nbr

I used patient_nbr only for splitting the data, not as a model feature.

- patient_nbr identifies the same patient across multiple hospital encounters.
Some patients have multiple encounters, so the same patient could otherwise appear in both training and testing data.
This can cause data leakage and give the model an unrealistically good result.
## Why GroupShuffleSplit?

GroupShuffleSplit keeps all records belonging to the same patient in one group.

### An encounter represents one hospital visit or admission of a patient. A patient can have multiple encounters, I use patient_nbr as the grouping variable during train-test splitting.

```python
# Identify potential outliers in numerical features

numeric_features = [
    "number_outpatient",
    "number_emergency",
    "number_inpatient"
]

X_train_missing[numeric_features].describe()
```

```python
# Identify potential outliers using the IQR method

Q1 = X_train_missing[numeric_features].quantile(0.25)
Q3 = X_train_missing[numeric_features].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outlier_summary = pd.DataFrame({
    "Q1": Q1,
    "Q3": Q3,
    "IQR": IQR,
    "Lower_Bound": lower_bound,
    "Upper_Bound": upper_bound
})

outlier_summary
```

### I used the IQR method to identify potential outliers in the numerical utilization features. The variables are highly right-skewed, and because Q1 and Q3 are both zero for outpatient and emergency visits, IQR gives a zero boundary. These extreme values can represent genuine patient utilization history rather than errors, so I decided not to remove them. For inpatient visits, values above 2.5 are statistically flagged, but I also retain them because they can represent legitimate previous hospitalizations.

```python
# Identify potential outliers in y_train using IQR

Q1_y_train = y_train.quantile(0.25)
Q3_y_train = y_train.quantile(0.75)

IQR_y_train = Q3_y_train - Q1_y_train

lower_y_train = Q1_y_train - 1.5 * IQR_y_train
upper_y_train = Q3_y_train + 1.5 * IQR_y_train

y_train_outlier_summary = pd.DataFrame({
    "Q1": [Q1_y_train],
    "Q3": [Q3_y_train],
    "IQR": [IQR_y_train],
    "Lower_Bound": [lower_y_train],
    "Upper_Bound": [upper_y_train]
})

y_train_outlier_summary
```

```python
# Identify potential outliers in y_test using IQR

Q1_y_test = y_test.quantile(0.25)
Q3_y_test = y_test.quantile(0.75)

IQR_y_test = Q3_y_test - Q1_y_test

lower_y_test = Q1_y_test - 1.5 * IQR_y_test
upper_y_test = Q3_y_test + 1.5 * IQR_y_test

y_test_outlier_summary = pd.DataFrame({
    "Q1": [Q1_y_test],
    "Q3": [Q3_y_test],
    "IQR": [IQR_y_test],
    "Lower_Bound": [lower_y_test],
    "Upper_Bound": [upper_y_test]
})

y_test_outlier_summary
```

## age_group
- Our dataset's age column is already provided as age ranges, rather than exact ages. We can create a broader age_group feature to reduce the number     of categories and give the model a simpler age-based grouping.

```python
# Convert the original age ranges into broader age groups

def create_age_group(age_range):
    if age_range in ["[0-10)", "[10-20)", "[20-30)", "[30-40)"]:
        return "Young_Adult"
    elif age_range in ["[40-50)", "[50-60)"]:
        return "Middle_Age"
    elif age_range in ["[60-70)", "[70-80)"]:
        return "Senior"
    elif age_range in ["[80-90)", "[90-100)"]:
        return "Older"
    else:
        return "Unknown"
```

```python
# Create age_group for training data

X_train_fe["age_group"] = X_train_fe["age"].apply(create_age_group)
```

```python
# Create age_group for testing data

X_test_fe["age_group"] = X_test_fe["age"].apply(create_age_group)
```

### Create the remaining engineered features
- diag_1_category
- diag_2_category
- diag_3_category

```python
# Convert ICD-9-style diagnosis codes into broad clinical categories

def create_diagnosis_category(code):
    code = str(code)

    if code == "Missing":
        return "Missing"

    if code.startswith("250"):
        return "Diabetes"

    if code[0].isdigit():
        code_num = float(code)

        if 390 <= code_num <= 459:
            return "Circulatory"

        elif 460 <= code_num <= 519:
            return "Respiratory"

        elif 520 <= code_num <= 579:
            return "Digestive"

        elif 580 <= code_num <= 629:
            return "Genitourinary"

        elif 630 <= code_num <= 679:
            return "Pregnancy"

        elif 680 <= code_num <= 709:
            return "Skin"

        elif 710 <= code_num <= 739:
            return "Musculoskeletal"

        elif 800 <= code_num <= 999:
            return "Injury"

        elif 140 <= code_num <= 239:
            return "Neoplasms"

    if code.startswith("E"):
        return "External_Cause"

    if code.startswith("V"):
        return "Supplementary"

    return "Other"
```

```python
# Create diagnosis category features for training data

X_train_fe["diag_1_category"] = X_train_fe["diag_1"].apply(
    create_diagnosis_category
)

X_train_fe["diag_2_category"] = X_train_fe["diag_2"].apply(
    create_diagnosis_category
)

X_train_fe["diag_3_category"] = X_train_fe["diag_3"].apply(
    create_diagnosis_category
)
```

```python
# Create diagnosis category features for testing data

X_test_fe["diag_1_category"] = X_test_fe["diag_1"].apply(
    create_diagnosis_category
)

X_test_fe["diag_2_category"] = X_test_fe["diag_2"].apply(
    create_diagnosis_category
)

X_test_fe["diag_3_category"] = X_test_fe["diag_3"].apply(
    create_diagnosis_category
)
```

```python
numerical_candidates = [
    "number_outpatient",
    "number_emergency",
    "number_inpatient",
    "prior_utilization",
    "emergency_history_flag",
    "inpatient_history_flag"
]

categorical_candidates = [
    "age",
    "gender",
    "race",
    "admission_type_id",
    "admission_source_id",
    "medical_specialty",
    "diag_1",
    "diag_2",
    "diag_3",
    "payer_code",
    "age_group",
    "diag_1_category",
    "diag_2_category",
    "diag_3_category",
    "admission_source_group",
    "admission_type_group"
]
```

```python
# Calculate numerical feature relevance using squared correlation

numerical_correlation = (
    X_train_fe[numerical_candidates]
    .corrwith(y_train)
)

numerical_relevance = (
    numerical_correlation
    .pow(2)
    .sort_values(ascending=False)
    .to_frame(name="Effect_Size")
)

numerical_relevance
```

### Categorical Feature Selection
## For categorical X variables and continuous Y, we can use ANOVA F-test directly on the original categories.

```python
from scipy.stats import f_oneway

categorical_results = []

for feature in categorical_candidates:

    temp = pd.DataFrame({
        "feature": X_train_fe[feature],
        "target": y_train
    })

    groups = [
        group["target"].values
        for _, group in temp.groupby("feature")
        if len(group) > 1
    ]

    if len(groups) >= 2:
        f_stat, p_value = f_oneway(*groups)

        grand_mean = temp["target"].mean()

        group_stats = (
            temp.groupby("feature")["target"]
            .agg(["mean", "count"])
        )

        ss_between = (
            group_stats["count"]
            * (group_stats["mean"] - grand_mean) ** 2
        ).sum()

        ss_total = (
            (temp["target"] - grand_mean) ** 2
        ).sum()

        eta_squared = ss_between / ss_total

        categorical_results.append({
            "Feature": feature,
            "F_Score": f_stat,
            "P_Value": p_value,
            "Eta_Squared": eta_squared
        })
```

```python
categorical_relevance = (
    pd.DataFrame(categorical_results)
    .sort_values(
        "Eta_Squared",
        ascending=False
    )
)

categorical_relevance
```

### Now combine the numerical and categorical results into one table.

```python
numerical_results = numerical_relevance.reset_index()

numerical_results.columns = [
    "Feature",
    "Effect_Size"
]

numerical_results["Feature_Type"] = "Numerical"

categorical_results_final = categorical_relevance[
    ["Feature", "Eta_Squared"]
].copy()

categorical_results_final.columns = [
    "Feature",
    "Effect_Size"
]

categorical_results_final["Feature_Type"] = "Categorical"

feature_selection_results = pd.concat(
    [
        numerical_results,
        categorical_results_final
    ],
    ignore_index=True
)

feature_selection_results = (
    feature_selection_results
    .sort_values("Effect_Size", ascending=False)
    .reset_index(drop=True)
)

feature_selection_results
```

```python
# Bias-Variance Study for Linear Regression Models

linear_models = {
    "Linear Regression": {
        "train_rmse": lr_train_rmse,
        "test_rmse": lr_test_rmse,
        "train_r2": lr_train_r2,
        "test_r2": lr_test_r2
    },
    "Polynomial Regression": {
        "train_rmse": poly_train_rmse,
        "test_rmse": poly_test_rmse,
        "train_r2": poly_train_r2,
        "test_r2": poly_test_r2
    },
    "Lasso Regression": {
        "train_rmse": lasso_train_rmse,
        "test_rmse": lasso_test_rmse,
        "train_r2": lasso_train_r2,
        "test_r2": lasso_test_r2
    },
    "Ridge Regression": {
        "train_rmse": ridge_train_rmse,
        "test_rmse": ridge_test_rmse,
        "train_r2": ridge_train_r2,
        "test_r2": ridge_test_r2
    }
}

linear_results = []

for model_name, metrics in linear_models.items():

    rmse_gap = metrics["test_rmse"] - metrics["train_rmse"]
    r2_gap = metrics["train_r2"] - metrics["test_r2"]

    # Practical Bias-Variance interpretation
    if metrics["train_r2"] < 0.30 and r2_gap < 0.10:
        bias = "High"
        variance = "Low"
        fit = "Underfitting"

    elif r2_gap >= 0.15:
        bias = "Low"
        variance = "High"
        fit = "Overfitting"

    else:
        bias = "Balanced"
        variance = "Moderate"
        fit = "Good Fit"

    linear_results.append({
        "Model": model_name,
        "Train RMSE": metrics["train_rmse"],
        "Test RMSE": metrics["test_rmse"],
        "Train R²": metrics["train_r2"],
        "Test R²": metrics["test_r2"],
        "RMSE Gap": rmse_gap,
        "R² Gap": r2_gap,
        "Bias": bias,
        "Variance": variance,
        "Fit": fit
    })

linear_bias_variance_df = pd.DataFrame(linear_results)

linear_bias_variance_df
```

### Step 1C — Categorical Feature Signal

- We'll measure how much the average LOS differs between categories using η² (eta-squared).

- We have a continuous target (time_in_hospital) and categorical predictors. Eta-squared tells us what proportion of target variation is associated with differences between categories.

```python
# ============================================================
# 1.4 IMPROVE MODEL PERFORMANCE
# Step 1C: Categorical Feature Signal
# ============================================================

import pandas as pd
from scipy.stats import f_oneway


# ------------------------------------------------------------
# Categorical features in the final selected feature set
# ------------------------------------------------------------

categorical_features = [
    "age",
    "gender",
    "race",
    "admission_type_id",
    "admission_source_id",
    "medical_specialty",
    "diag_1",
    "diag_2",
    "diag_3",
    "payer_code",
    "diag_1_category",
    "diag_2_category",
    "diag_3_category",
    "admission_source_group"
]


categorical_relationship = []


# ------------------------------------------------------------
# Calculate ANOVA F-statistic and Eta Squared
# ------------------------------------------------------------

for feature in categorical_features:

    grouped_values = [
        group["target"].values
        for _, group in (
            pd.DataFrame({
                feature: X_train_mod[feature],
                "target": y_train
            })
            .groupby(feature)
        )
    ]

    f_statistic, p_value = f_oneway(*grouped_values)

    grand_mean = y_train.mean()

    between_group_ss = sum(
        len(group) * (group["target"].mean() - grand_mean) ** 2
        for _, group in (
            pd.DataFrame({
                feature: X_train_mod[feature],
                "target": y_train
            })
            .groupby(feature)
        )
    )

    total_ss = ((y_train - grand_mean) ** 2).sum()

    eta_squared = between_group_ss / total_ss

    categorical_relationship.append({
        "Feature": feature,
        "F-Statistic": f_statistic,
        "P-Value": p_value,
        "Eta Squared": eta_squared
    })


categorical_relationship_df = (
    pd.DataFrame(categorical_relationship)
    .sort_values(
        "Eta Squared",
        ascending=False
    )
    .reset_index(drop=True)
)


categorical_relationship_df.round(4)
```

```python
# Step 1E.1 — Analyze nonlinear patterns in key numerical features

import pandas as pd

nonlinear_features = [
    "number_outpatient",
    "number_emergency",
    "number_inpatient",
    "prior_utilization"
]

nonlinear_analysis = []

for feature in nonlinear_features:

    analysis_df = pd.DataFrame({
        "feature": X_train_mod[feature],
        "target": y_train
    })

    if feature == "number_outpatient":
        analysis_df["group"] = pd.cut(
            analysis_df["feature"],
            bins=[-1, 0, 1, 3, 5, float("inf")],
            labels=["0", "1", "2-3", "4-5", "6+"]
        )

    elif feature == "number_emergency":
        analysis_df["group"] = pd.cut(
            analysis_df["feature"],
            bins=[-1, 0, 1, 2, 4, float("inf")],
            labels=["0", "1", "2", "3-4", "5+"]
        )

    elif feature == "number_inpatient":
        analysis_df["group"] = pd.cut(
            analysis_df["feature"],
            bins=[-1, 0, 1, 2, 4, float("inf")],
            labels=["0", "1", "2", "3-4", "5+"]
        )

    elif feature == "prior_utilization":
        analysis_df["group"] = pd.cut(
            analysis_df["feature"],
            bins=[-1, 0, 1, 3, 5, float("inf")],
            labels=["0", "1", "2-3", "4-5", "6+"]
        )

    grouped_result = (
        analysis_df
        .groupby("group", observed=False)["target"]
        .agg(
            Count="count",
            Mean_LOS="mean",
            Median_LOS="median"
        )
        .reset_index()
    )

    grouped_result.insert(0, "Feature", feature)

    nonlinear_analysis.append(grouped_result)

nonlinear_analysis_df = pd.concat(
    nonlinear_analysis,
    ignore_index=True
)

nonlinear_analysis_df.round(3)
```

```python
# Step 1F — Controlled Decision Tree Regression
# Purpose: Test whether restricting tree complexity improves generalization

from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np
import pandas as pd

controlled_tree = DecisionTreeRegressor(
    max_depth=8,
    min_samples_leaf=50,
    min_samples_split=100,
    random_state=42
)

controlled_tree.fit(X_train_final, y_train)

y_train_pred_controlled_tree = controlled_tree.predict(X_train_final)
y_test_pred_controlled_tree = controlled_tree.predict(X_test_final)

controlled_tree_train_rmse = np.sqrt(
    mean_squared_error(y_train, y_train_pred_controlled_tree)
)

controlled_tree_test_rmse = np.sqrt(
    mean_squared_error(y_test, y_test_pred_controlled_tree)
)

controlled_tree_train_r2 = r2_score(
    y_train,
    y_train_pred_controlled_tree
)

controlled_tree_test_r2 = r2_score(
    y_test,
    y_test_pred_controlled_tree
)

controlled_tree_rmse_gap = (
    controlled_tree_test_rmse -
    controlled_tree_train_rmse
)

controlled_tree_r2_gap = (
    controlled_tree_train_r2 -
    controlled_tree_test_r2
)

controlled_tree_results = pd.DataFrame({
    "Model": ["Controlled Decision Tree"],
    "Train RMSE": [controlled_tree_train_rmse],
    "Test RMSE": [controlled_tree_test_rmse],
    "Train R2": [controlled_tree_train_r2],
    "Test R2": [controlled_tree_test_r2],
    "RMSE Gap": [controlled_tree_rmse_gap],
    "R2 Gap": [controlled_tree_r2_gap]
})

controlled_tree_results.round(4)
```

```python
## 1.4.1 — Polynomial Regression tuning
# 1.4.1 — Polynomial Regression: compare degrees

from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np
import pandas as pd

polynomial_results = []

for degree in [1, 2, 3]:

    polynomial_transformer = PolynomialFeatures(
        degree=degree,
        include_bias=False
    )

    X_train_poly_num = polynomial_transformer.fit_transform(
        X_train_poly_base
    )

    X_test_poly_num = polynomial_transformer.transform(
        X_test_poly_base
    )

    # Combine polynomial numerical features
    # with the existing categorical representation.
    X_train_poly = np.hstack([
        X_train_cat_encoded_df.values,
        X_train_poly_num,
        X_train_mod[["inpatient_history_flag"]].values
    ])

    X_test_poly = np.hstack([
        X_test_cat_encoded_df.values,
        X_test_poly_num,
        X_test_mod[["inpatient_history_flag"]].values
    ])

    polynomial_model = Ridge(alpha=1.0)

    polynomial_model.fit(
        X_train_poly,
        y_train
    )

    y_train_pred = polynomial_model.predict(X_train_poly)
    y_test_pred = polynomial_model.predict(X_test_poly)

    train_rmse = np.sqrt(
        mean_squared_error(y_train, y_train_pred)
    )

    test_rmse = np.sqrt(
        mean_squared_error(y_test, y_test_pred)
    )

    train_r2 = r2_score(
        y_train,
        y_train_pred
    )

    test_r2 = r2_score(
        y_test,
        y_test_pred
    )

    polynomial_results.append({
        "Degree": degree,
        "Train RMSE": train_rmse,
        "Test RMSE": test_rmse,
        "Train R2": train_r2,
        "Test R2": test_r2,
        "RMSE Gap": test_rmse - train_rmse,
        "R2 Gap": train_r2 - test_r2
    })

polynomial_tuning_results = pd.DataFrame(
    polynomial_results
)

polynomial_tuning_results.round(4)
```

```python
# Best SVM Hyperparameters

best_svm_params = svm_grid_search.best_params_
best_svm_cv_rmse = -svm_grid_search.best_score_

pd.DataFrame({
    "Best_C": [best_svm_params["C"]],
    "Best_Epsilon": [best_svm_params["epsilon"]],
    "Best_CV_RMSE": [best_svm_cv_rmse]
})
```

## Final Tuned SVM Evaluation

- Now we can evaluate the selected configuration on the untouched test set.

 Hyperparameters were selected using only the training folds. This is the correct point to measure performance on unseen test data.

```python
# Train Final Tuned SVM

tuned_svm = LinearSVR(
    C=best_svm_params["C"],
    epsilon=best_svm_params["epsilon"],
    loss="squared_epsilon_insensitive",
    dual=False,
    max_iter=5000,
    random_state=42
)

tuned_svm.fit(
    X_train_svm,
    y_train
)
```

 - we are we tuning Random Forest?

- Test R² = 0.1640
- Large R² gap = 0.7181

So our tuning objective is to reduce variance/overfitting while preserving the Random Forest's nonlinear learning ability. We will use the full training dataset and training-only 5-fold CV—no sampling and no test-set tuning.

```python
# Reuse the existing sparse representation
X_train_rf = X_train_ridge
X_test_rf = X_test_ridge
```

## Random Forest hyperparameter grid

- For an overfitting Random Forest, we mainly control:

- max_depth → limits individual tree complexity
- min_samples_split → prevents very small splits
- min_samples_leaf → prevents highly specific leaves
- max_features → controls how many features each tree considers

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import GridSearchCV

rf_regressor = RandomForestRegressor(
    random_state=42,
    n_jobs=1
)

rf_param_grid = {
    "n_estimators": [200],
    "max_depth": [8, 12, 16],
    "min_samples_split": [20, 50, 100],
    "min_samples_leaf": [5, 10, 20],
    "max_features": ["sqrt", 0.5]
}

rf_grid_search = GridSearchCV(
    estimator=rf_regressor,
    param_grid=rf_param_grid,
    scoring="neg_root_mean_squared_error",
    cv=5,
    n_jobs=1,
    return_train_score=True,
    error_score="raise"
)

rf_grid_search.fit(X_train_rf, y_train)
```

```python
rf_cv_results = pd.DataFrame(
    rf_grid_search.cv_results_
)

rf_results_table = (
    rf_cv_results[
        [
            "param_max_depth",
            "param_min_samples_split",
            "param_min_samples_leaf",
            "param_max_features",
            "mean_train_score",
            "mean_test_score",
            "std_test_score"
        ]
    ]
    .copy()
)

rf_results_table["Train_RMSE"] = (
    -rf_results_table["mean_train_score"]
)

rf_results_table["CV_RMSE"] = (
    -rf_results_table["mean_test_score"]
)

rf_results_table = rf_results_table[
    [
        "param_max_depth",
        "param_min_samples_split",
        "param_min_samples_leaf",
        "param_max_features",
        "Train_RMSE",
        "CV_RMSE",
        "std_test_score"
    ]
].sort_values("CV_RMSE")

rf_results_table.head(15)
```

```python
## Best Random Forest parameters
best_rf_params = rf_grid_search.best_params_

best_rf_cv_rmse = -rf_grid_search.best_score_

pd.DataFrame({
    "Best_Max_Depth": [best_rf_params["max_depth"]],
    "Best_Min_Samples_Split": [
        best_rf_params["min_samples_split"]
    ],
    "Best_Min_Samples_Leaf": [
        best_rf_params["min_samples_leaf"]
    ],
    "Best_Max_Features": [
        best_rf_params["max_features"]
    ],
    "Best_CV_RMSE": [best_rf_cv_rmse]
})
```

## Final Tuned Random Forest Evaluation

- Now fit the best configuration on all 81,613 training rows, then evaluate on the untouched 20,153 test rows.

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np

# Train Random Forest using the best hyperparameters
tuned_rf_regressor = RandomForestRegressor(
    n_estimators=200,
    max_depth=16,
    min_samples_split=20,
    min_samples_leaf=5,
    max_features=0.5,
    random_state=42,
    n_jobs=1
)

tuned_rf_regressor.fit(
    X_train_rf,
    y_train
)

# Generate predictions
y_train_pred_tuned_rf = tuned_rf_regressor.predict(X_train_rf)
y_test_pred_tuned_rf = tuned_rf_regressor.predict(X_test_rf)

# RMSE
tuned_rf_train_rmse = np.sqrt(
    mean_squared_error(y_train, y_train_pred_tuned_rf)
)

tuned_rf_test_rmse = np.sqrt(
    mean_squared_error(y_test, y_test_pred_tuned_rf)
)

# R²
tuned_rf_train_r2 = r2_score(
    y_train,
    y_train_pred_tuned_rf
)

tuned_rf_test_r2 = r2_score(
    y_test,
    y_test_pred_tuned_rf
)

# Generalization gaps
tuned_rf_rmse_gap = (
    tuned_rf_test_rmse - tuned_rf_train_rmse
)

tuned_rf_r2_gap = (
    tuned_rf_train_r2 - tuned_rf_test_r2
)

# Bias-variance classification
if tuned_rf_train_r2 < 0.5 and abs(tuned_rf_r2_gap) < 0.10:
    tuned_rf_bias = "High"
    tuned_rf_variance = "Low"
    tuned_rf_fit = "Underfitting"

elif tuned_rf_r2_gap > 0.15:
    tuned_rf_bias = "Low"
    tuned_rf_variance = "High"
    tuned_rf_fit = "Overfitting"

else:
    tuned_rf_bias = "Low"
    tuned_rf_variance = "Low"
    tuned_rf_fit = "Good Fit"

rf_final_results = pd.DataFrame({
    "Model": ["Tuned Random Forest"],
    "Train_RMSE": [tuned_rf_train_rmse],
    "Test_RMSE": [tuned_rf_test_rmse],
    "Train_R2": [tuned_rf_train_r2],
    "Test_R2": [tuned_rf_test_r2],
    "RMSE_Gap": [tuned_rf_rmse_gap],
    "R2_Gap": [tuned_rf_r2_gap],
    "Bias": [tuned_rf_bias],
    "Variance": [tuned_rf_variance],
    "Fit": [tuned_rf_fit]
})

rf_final_results
```

- Random Forest hyperparameter tuning successfully eliminated the severe overfitting of the baseline model, but the selected configuration over-regularized the forest and resulted in underfitting. It does not outperform the current leading models.

Transform the unseen encounter

Your saved preprocessor.joblib contains the One-Hot Encoder and Standard Scaler. We need to reproduce the same final 2,359-feature structure that XGBoost saw during training.

```python
# Transform the unseen encounter using the saved preprocessing artifacts

loaded_preprocessor = joblib.load(
    preprocessing_path
)

unseen_onehot_encoder = loaded_preprocessor["onehot_encoder"]
unseen_standard_scaler = loaded_preprocessor["standard_scaler"]

unseen_categorical_features = (
    loaded_preprocessor["categorical_features"]
)

unseen_scaled_numerical_features = (
    loaded_preprocessor["scaled_numerical_features"]
)

# Encode categorical features
unseen_cat_encoded = unseen_onehot_encoder.transform(
    unseen_encounter[unseen_categorical_features]
)

unseen_encoded_feature_names = (
    unseen_onehot_encoder.get_feature_names_out(
        unseen_categorical_features
    )
)

unseen_cat_encoded_df = pd.DataFrame(
    unseen_cat_encoded,
    columns=unseen_encoded_feature_names,
    index=unseen_encounter.index
)

# Scale numerical features
unseen_num_scaled = unseen_standard_scaler.transform(
    unseen_encounter[unseen_scaled_numerical_features]
)

unseen_num_scaled_df = pd.DataFrame(
    unseen_num_scaled,
    columns=unseen_scaled_numerical_features,
    index=unseen_encounter.index
)

# Combine transformed features
unseen_final = pd.concat(
    [
        unseen_cat_encoded_df,
        unseen_num_scaled_df,
        unseen_encounter[
            ["inpatient_history_flag"]
        ]
    ],
    axis=1
)

# Apply the same feature-name sanitization used during XGBoost training
import re

unseen_final.columns = [
    re.sub(
        r"[\[\]<>]",
        "_",
        str(column)
    )
    for column in unseen_final.columns
]

# Get the expected feature names from the training DataFrame
expected_feature_names = X_train_final.columns

# Ensure exactly the same feature order and feature set
unseen_final = unseen_final.reindex(
    columns=expected_feature_names,
    fill_value=0
)

# Verify the transformed shape
pd.DataFrame({
    "Rows": [unseen_final.shape[0]],
    "Features": [unseen_final.shape[1]],
    "Expected_Features": [X_train_final.shape[1]],
    "Missing_Values": [unseen_final.isna().sum().sum()]
})
```

### Generate the Unseen Prediction
- This is the actual inference step: we pass the unseen encounter through the saved final XGBoost model and obtain the predicted hospital length of stay. The true value (10 days) is used only afterward for comparison, not as an input.

```python
# Load the finalized XGBoost model

loaded_final_model = joblib.load(
    final_model_path
)

# Generate prediction for the unseen encounter
unseen_prediction = loaded_final_model.predict(
    unseen_final
)

predicted_los = float(
    unseen_prediction[0]
)

# Compare prediction with the actual unseen value
prediction_result = pd.DataFrame({
    "Encounter_Index": [unseen_index],
    "Predicted_Length_of_Stay_Days": [predicted_los],
    "Actual_Length_of_Stay_Days": [unseen_actual_los],
    "Absolute_Error_Days": [
        abs(predicted_los - unseen_actual_los)
    ]
})

prediction_result
```

## Deployment Validation — Notebook vs FastAPI

```python
import json
import urllib.request
import numpy as np


# ============================================================
# 1. Take ONE exact raw test row
# ============================================================

regression_raw_features = [
    "age",
    "gender",
    "race",
    "admission_type_id",
    "admission_source_id",
    "medical_specialty",
    "number_outpatient",
    "number_emergency",
    "number_inpatient",
    "diag_1",
    "diag_2",
    "diag_3",
    "payer_code"
]

regression_test_row = (
    X_test.iloc[0][regression_raw_features]
    .to_dict()
)

regression_payload = {
    key: (
        value.item()
        if hasattr(value, "item")
        else value
    )
    for key, value in regression_test_row.items()
}


# ============================================================
# 2. Notebook prediction
# ============================================================

notebook_prediction = float(
    final_xgb_regressor.predict(
        X_test_final.iloc[[0]]
    )[0]
)

notebook_prediction = float(
    np.clip(
        notebook_prediction,
        1,
        14
    )
)


# ============================================================
# 3. Send EXACT SAME row to FastAPI
# ============================================================

request = urllib.request.Request(
    "http://127.0.0.1:8000/predict/length-of-stay",
    data=json.dumps(
        regression_payload
    ).encode("utf-8"),
    headers={
        "Content-Type": "application/json"
    },
    method="POST"
)

with urllib.request.urlopen(request) as response:

    fastapi_result = json.loads(
        response.read().decode("utf-8")
    )

fastapi_prediction = float(
    fastapi_result["prediction"]
)


# ============================================================
# 4. Compare
# ============================================================

difference = abs(
    notebook_prediction -
    fastapi_prediction
)

match = np.isclose(
    notebook_prediction,
    fastapi_prediction,
    rtol=1e-6,
    atol=1e-6
)


# ============================================================
# 5. Final validation output
# ============================================================

print("=" * 60)
print("REGRESSION DEPLOYMENT VALIDATION")
print("=" * 60)

print(
    "Notebook prediction :",
    notebook_prediction
)

print(
    "FastAPI prediction  :",
    fastapi_prediction
)

print(
    "Absolute difference :",
    difference
)

print(
    "MATCH               :",
    match
)

print("\nExact JSON sent to FastAPI:")
print(
    json.dumps(
        regression_payload,
        indent=2
    )
)
```
